# Machine Learning–Driven Intention-Aware Smart Crosswalk
## Simulation Study & Equity Analysis

This notebook reproduces the core findings of the paper, detailing the trade-offs between heuristic and uncertainty-propagated ML control policies.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Data
df_normal = pd.read_csv('outputs/mc_results_normal.csv')
df_equity = pd.read_csv('outputs/equity_report.csv')
```

### 1. Performance Trade-offs
Comparing Vehicle Delay vs Pedestrian Wait Times across baselines.

```python
means = df_normal.groupby('Controller').mean(numeric_only=True)
sns.scatterplot(data=means, x='avg_delay', y='avg_wait', hue=means.index, s=200)
plt.title('Delay vs Wait Time')
plt.show()
```

### 2. Equity & Fairness Analysis
The Gap Ratio compares vulnerable pedestrian wait times to normal pedestrians (Target = 1.0).

```python
pivot = df_equity.pivot(index='Controller', columns='Pedestrian Profile', values='Gap Ratio')
sns.heatmap(pivot, annot=True, cmap='coolwarm', center=1.0)
plt.title('Equity Gap Ratio by Profile')
plt.show()
```
